```python
# Write a Python program to calculate the total marks and percentage from the given dictionary containing the marks of a student named Ali. 
# Based on the percentage, 
# display the result as Distinct if the percentage is 75% or above, 
# First Class if it is 60% or above but less than 75%, 
# Pass if it is 35% or above but less than 60%, 
# and Fail if it is below 35%.

ali= {
    "English": 42,
    "Mathematics": 38,
    "Science": 45,
    "Social Science": 40,
    "Computer": 48
}
```

```python
total_marks = 0
for marks in ali.values(): total_marks += marks
percentage = total_marks*100/(len(ali)*50)
print(percentage)

if percentage >= 75: print("Distinction")
elif percentage >= 60: print("First Class")
elif percentage >= 35: print("Pass")
else: print("Fail")
```

```python
def getResult(student_dict):
    percentage = sum(student_dict.values())*100/(len(student_dict)*50)
    
    if percentage >= 75: return "Distinction"
    elif percentage >= 60: return "First Class"
    elif percentage >= 35: return "Pass"
    else: return "Fail"
```

```python
getResult(ali)
```

```python
def getMaxNumber(lst):
    max = lst[0]
    for i in lst:
        if i > max: max = i

    return max
```

```python
getMaxNumber([4,12312,2123,1312312,555,3231,3333333])
```

```python
books=[
    {"id": 1, "name":'Sahih al-Bukhari',"author":'Imam Bukhari'},
    {"id": 2, "name":'Sahih al-Muslim','author':'Imam Muslim'},
    {"id": 3, "name":'Miskat','author':'Waliuddin'}
]
```

```python
def get_books():
    return books
```

```python
get_books()
```

```python
def get_book(id):
    for book in books:
        if book["id"] == id: return book
    return {"message": "Book not found"}
```

```python
get_book(2)
```

```python
get_book(5)
```

```python
def create_book(name, author):
    book_id = len(books) + 1
    books.append({"id": book_id, "name": name, "author": author})
    return {"message": "Book created successfully!"}
```

```python
create_book("ABC", "DEF")
```

```python
create_book("Hello Python", "Tom")
```

```python
get_books()
```

```python
def update_book(id, name = None, author = None):
    for book in books:
        if book["id"] == id:
            if name:
                book["name"] = name
            if author:
                book["author"] = author
            return {"message": "Book updated successfully"}
    return {"message": "Book Not Found!"}
```

```python
update_book(2, "Java", "Tim")
```

```python
get_books()
```

```python
def delete_book(id):
    for book in books:
        if book["id"] == id:
            books.remove(book)
            return {"message": "Book Deleted"}
    return {"message": "Book not found"}
```

```python
delete_book(2)
```

```python
get_books()
```
